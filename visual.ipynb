```python
import matplotlib.pyplot as plt

lengths = [
    len(data["input_ids"])
    for data in train_tokenized_dataset
]

bins = [0, 100, 200, 300, 500, 700, 1000, float("inf")]

labels = [
    "0–100", "101–200", "201–300",
    "301–500", "501–700", "701–1000", "1000+"
]

counts = [
    sum(lower < length <= upper for length in lengths)
    for lower, upper in zip(bins[:-1], bins[1:])
]

plt.figure(figsize=(10, 5))
plt.bar(labels, counts)
plt.xlabel("Sequence length range (tokens)")
plt.ylabel("Number of examples")
plt.title("Sequence Length Sparsity")
plt.xticks(rotation=30)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```
